# Задача 1

Пусть $X_1, X_2, \ldots, X_n$ — выборка из экспоненциального распределения с параметром $\lambda$. Найти оценку максимального правдоподобия параметра $\lambda$, сравнить ее с байесовской оценкой (MAP и математическое ожидание апостреорного распределения), подобрав сопряженное распределение. Сравнить полученные байесовские оценки с оценкой MLE. Найти предсказательное распределение

# Формула, которая описывает экспоненциальное распределение
$$f(x_i \mid \lambda) = \lambda e^{\lambda x_i}$$
$\lambda$ будет являться показателем интенсивности событий

Обозначем все суммы какой-то выборки X как
$$S= \sum_{i=1}^{n}X_i$$

## Оценка MLE
Функция правдоподобия L($\lambda$)
$$L(\lambda) = \prod_{i=1}^{n} f(x_i \mid \lambda) = \lambda^n e^{-\lambda S}$$
Это функция, которая показывает насколько вероятно получить конкретный набор данных $X_1,...,X_n$, если истинный параметр скорости был равен $\lambda$
Чтобы найти такой $\lambda$, я должен максимизировать эту функцию правдоподобия \
Перезаписываю функцию L($\lambda$) логорифмируя обе части $\ln(L(\lambda)) = n\ln \lambda - \lambda S$. Далее обозначу функцию $\ln(L(\lambda))$ как функцию $\ell(\lambda)$. Так как логорифм является монотонно возрастающей функция если я найду максимум от функции $\ell(\lambda)$, то я найду и максимум от функции L($\lambda$).  
Возьму производную от функции $\ell(\lambda)$ и приравняю ее к нулю. 
$$\frac{d\ell}{d\lambda} = \frac{n}{\lambda} - S = 0 \hspace{1pt} \implies \hat{\lambda}_{\text{MLE}} = \frac{n}{S}$$
Это единственное значение $\lambda$, которое делает выборку наиболее вероятной согласно математической модели.  
Найду математическое ожидание от значения $\lambda$, таким образоом, я смогу понять смещена ли оценка $\hat{\lambda}_{\text{MLE}}$ или нет.
$$E\left[\hat{\lambda}_{\text{MLE}}\right] = \lambda\times \frac{n}{n-1}$$
Для того, чтобы найти это занчение я предварительно нашел $E\left[\frac{1}{S}\right]$и потом это значение необходимо будет умножить на n. 
$$E[g(S)] = \int_{-\infty}^{\infty} g(s) f_S(s) \, ds$$
$$E\left[\frac{1}{S}\right] = \int_{0}^{\infty}\frac{1}{S}\cdot\frac{\lambda^n}{\Gamma(n)}s^{n-1}e^{\lambda s}ds$$
$$E\left[\frac{1}{S}\right] = \frac{\lambda^n}{\Gamma(n)} \cdot \frac{\Gamma(n-1)}{\lambda^{n-1}}$$

$$E\left[\frac{1}{S}\right] = \frac{\lambda^{n-(n-1)}}{\Gamma(n)} \Gamma(n-1)$$

$$E\left[\frac{1}{S}\right] = \frac{\lambda \Gamma(n-1)}{\Gamma(n)}$$
Так как $\frac{n}{n-1} > 1 $ я получил смещенную оценку.  
Провведем коррекцию смещенной оценки. Я хочу, чтобы $E[\tilde{\lambda}] = \lambda$ Умножу $\hat{\lambda}_\text{MLE} \text{на поправочный коэффициент} \frac{n-1}{n}$. Таким образом, я получаю $\tilde{\lambda} = \frac{n-1}{S}$.  
Вывод: теперь я получаю несмещенную оценку для устранения систематической ошибки. 
# Получение Априора и Апостериора
$$\pi(\lambda) = \Gamma(\lambda | \alpha,\beta) = \frac{\beta^\alpha}{\Gamma(\alpha)}\lambda^{\alpha-1}e^{-\beta\lambda}$$
Где $\alpha$ параметр формы, а $\beta$ параметр скорости.  
$E[\lambda]$: $\alpha / \beta$  
$\operatorname{Var}[\lambda]$: $\alpha / \beta^2$

Форма функции правдоподобия (Likelihood):$$L(\lambda) \propto \lambda^n e^{-\lambda S}$$Обратите внимание на то, как $\lambda$ и $e^{-\lambda}$ появляются в формуле.
Форма Гамма-априора $\pi(\lambda)$:$$\pi(\lambda) \propto \lambda^{\alpha-1} e^{-\beta\lambda}$$
Формула Гамма-распределения$$\pi(\lambda)=\mathrm{Gamma}(\lambda\mid\alpha,\beta)=\frac{\beta^{\alpha}}{\Gamma(\alpha)}\lambda^{\alpha-1} e^{-\beta\lambda},$$
Если бы вы продолжили байесовский анализ, вы бы вычислили апостериорное распределение $\pi(\lambda \mid \text{Данные})$ по формуле Байеса:$$\pi(\lambda \mid \mathbf{X}) \propto L(\lambda) \cdot \pi(\lambda)$$
$$\pi(\lambda \mid \mathbf{X}) \propto (\lambda^n e^{-\lambda S}) \cdot (\lambda^{\alpha-1} e^{-\beta\lambda})$$
Сопряженный априор (conjugate prior) — это априорное распределение, которое, будучи умножено на функцию правдоподобия, дает апостериорное распределение того же семейства.
В данном случае:Правдоподобие (Экспоненциальное) $\propto \lambda^{\text{константа}} e^{-\lambda \cdot \text{константа}}$.Гамма-априор $\propto \lambda^{\text{константа}-1} e^{-\lambda \cdot \text{константа}}$.
$$\pi(\lambda \mid \mathbf{X}) = \operatorname{Gamma}(\lambda \mid \alpha + n, \beta + S)$$
# Получение точных оценок из апостериорного распределения
Апостериорное среднее  
$E\left[\lambda \mid X \right] = \frac{\alpha+n}{\beta+S}$  
MAP  
$\hat{\lambda}_{MAP} = \frac{\alpha + n -1}{\beta + S}$  
Вывод, который можно сделать из двух основных Байесовских точечных оценок ($\hat\lambda_{\text{MAP}}$ и $E[\lambda \mid X]$), заключается в следующем:Байесовские точечные оценки — это разумный компромисс между вашей предварительной уверенностью (априором) и информацией, полученной из данных (правдоподобием). Они обеспечивают более устойчивую и стабильную оценку, чем классическая MLE, особенно при малом объеме данных.
# Сравнение Байесовских оценок с MLE
Запмшу $\hat{\lambda}_\text{MLE} = \frac{n}{S}$, $E\left[\lambda \mid X \right] = \frac{\alpha+n}{\beta+S}$, $\hat{\lambda}_{MAP} = \frac{\alpha + n -1}{\beta + S}$
По мере роста объема данных ($n \to \infty$ и $S \to \infty$):Влияние $\alpha$ и $\beta$ становится пренебрежимо малым.$\hat\lambda_{\text{MAP}} \approx \frac{n}{S}$ $$E[\lambda \mid X] \approx \frac{n}{S}$$ $$\hat\lambda_{\text{MLE}} \approx \frac{n}{S}$$ Все три оценки сходятся к одному и тому же значению. Это подтверждает, что если у вас достаточно данных, то неважно, с каких убеждений $(\alpha, \beta)$ будет предложено начать.
# Предсказательное распределение для нового наблюдения $X_n+1$
$$p(x_{n+1} \mid \mathbf{X}) = \int_{0}^{\infty} \underbrace{p(x_{n+1} \mid \lambda)}_{\text{Вероятность нового } X \text{ при данном } \lambda} \cdot \underbrace{p(\lambda \mid \mathbf{X})}_{\text{Наше знание о } \lambda \text{ после данных}} \, d\lambda$$
орма Распределения: Ломакс (Парето II)Формула:$$p(x_{n+1} \mid \mathbf{X}) = \underbrace{a' b'^{a'}}_{\text{Константа}} \cdot \frac{1}{(b'+x_{n+1})^{a'+1}}$$
Вывод: Тяжелые ХвостыТяжелый хвост означает, что, по сравнению с обычным экспоненциальным распределением, вероятность очень долгих времен $X_{n+1}$ выше, чем могли бы мы ожидать. Это логично: поскольку мы не уверены в $\lambda$, мы должны быть готовы к тому, что истинное $\lambda$ может быть очень маленьким, что приведет к очень долгим временам $X_{n+1}$.
Предсказательные Моменты (E[$\mathbf{X_{n+1}}$] и $\operatorname{Var}(\mathbf{X_{n+1}}$))A. Апостериорное Предсказательное Среднее $E[X_{n+1} \mid \mathbf{X}]$Формула:$$E[X_{n+1} \mid X] = \frac{b'}{a'-1} = \frac{\beta+S}{\alpha+n-1}$$Вывод:Это среднее предсказанное время работы следующего объекта. Оно равно математическому ожиданию обратной величины $\frac{1}{\lambda}$ по апостериорному распределению $p(\lambda \mid \mathbf{X})$.$E[X_{n+1} \mid X] \approx \frac{S}{n}$ (при больших $n$): При большом объеме данных предсказанное среднее время сходится к обратному MLE, $\frac{1}{\hat\lambda_{\text{MLE}}}$.B. Апостериорная Предсказательная Дисперсия $\operatorname{Var}(X_{n+1} \mid \mathbf{X})$Формула:$$\operatorname{Var}(X_{n+1} \mid X)=\frac{b'^2 a'}{(a'-1)^2(a'-2)}$$Вывод:Эта дисперсия показывает общую неопределенность вашего прогноза. Она всегда больше, чем дисперсия, которую вы получили бы, если бы просто подставили точечную оценку $\hat\lambda$ в формулу дисперсии экспоненциального распределения. Это различие вызвано тем, что:$$\operatorname{Var}(X_{n+1} \mid \mathbf{X}) = \underbrace{E[\operatorname{Var}(X_{n+1} \mid \lambda)]}_{\text{Случайность данных}} + \underbrace{\operatorname{Var}(E[X_{n+1} \mid \lambda])}_{\text{Неопределенность параметра } \lambda}$$То есть, общая неопределенность (дисперсия) — это сумма случайности самих данных и неопределенности оценки $\lambda$.

# Задача 2

**Мультиномиальное распределение**

Пусть проводится серия из $n$ испытаний и в результате каждого испытания происходит ровно одно событие из набора $A_1, A_2, \dots, A_m$, причем вероятности этих событий равны соответственно $\mathsf{p}_1, \mathsf{p}_2, \dots, \mathsf{p}_m$, причем
$$\sum_{i=1}^{m}\mathsf{p}_i = 1.$$

Тогда совместное распределение величин $X_1, X_2, \dots, X_m$, где $X_k$ — число наступлений события $A_k$ в серии из $n$ испытаний, задается вероятностями

$$
\mathsf{P}\left(X_1 = n_1, \dots, X_m = n_m, \right) = \frac{n!}{n_1!\dots n_m!}\mathsf{p}_1^{n_1}\dots \mathsf{p}_m^{n_m},
$$

где $n_1, n_2, \dots, n_m$ — произвольный набор целых неотрицательных чисел, таких что

$$\sum_{i=1}^m n_i = n.$$

Произведите байесовский вывод для мультиномиального распределения: найдите апостериорное распределение, используя в качестве сопоряженного распределения к правдоподобию [распределение Дирихле](https://ru.wikipedia.org/wiki/%D0%A0%D0%B0%D1%81%D0%BF%D1%80%D0%B5%D0%B4%D0%B5%D0%BB%D0%B5%D0%BD%D0%B8%D0%B5_%D0%94%D0%B8%D1%80%D0%B8%D1%85%D0%BB%D0%B5), найдите предсказательное распределение. Объясните результат.

# Модель и правдоподобие
Пусть в каждом испытании происходит ровно одно из событий $A_1,...,A_m$ с веротностями
$$p = (p_1,...,p_m), \hspace{20pt} p_i>=0, \hspace{20pt} \sum_{i=1}^{m} p_i = 1$$
После n испытаний наблюдаем счетчики
$$ X = (X_1,...,X_m), \hspace{20pt} X_i = n_i, \hspace{20pt} \sum_{i=1}^{m} n_i = n$$
ункция Правдоподобия (Likelihood)
$$p(\mathbf{n} \mid \mathbf{p}) = L(\mathbf{p}) = \underbrace{\frac{n!}{\prod_{i=1}^m n_i!}}_{\text{I. Мультиномиальный коэффициент}} \cdot \underbrace{\prod_{i=1}^m p_i^{\,n_i}}_{\text{II. Вероятность последовательности исходов}}$$
Функция правдоподобия Мультиномиального распределения $L(\mathbf{p})$ позволяет нам, наблюдая счетчики $\mathbf{n}$, оценить, насколько вероятна эта наблюдаемая комбинация при различных наборах вероятностей $\mathbf{p} = (p_1, \dots, p_m)$. Это основа для нахождения MLE или Байесовских оценок для векторов вероятностей $\mathbf{p}$.

# Споряженный априор: распределение Дирихле
Распределение Дирихле — это многомерное обобщение Бета-распределения. Оно является распределением вероятностей на наборе векторов вероятностей $\mathbf{p} = (p_1, \dots, p_m)$.
$\operatorname{Dir}(\boldsymbol{\alpha})$ описывает, насколько мы уверены в том, что истинный вектор вероятностей $\mathbf{p}$ выглядит именно так, а не иначе. Если $\alpha_1$ велико относительно других $\alpha_i$, это говорит о вашем убеждении, что $p_1$ должно быть большим.  
Я использую распоряжение Дирихле, потому что оно является сопряженным относительно мультиномиального правдоподобия.
Форма Мультиномиального Правдоподобия $L(\mathbf{p})$:$$L(\mathbf{p} \mid \mathbf{n}) \propto \prod_{i=1}^m p_i^{\,n_i}$$(Зависит от $p_i$ в виде произведения $p_i$ в степени $n_i$).Форма Дирихле-Априора $\pi(\mathbf{p})$:$$\pi(\mathbf{p}) \propto \prod_{i=1}^m p_i^{\,\alpha_i-1}$$

# Апостериорное распределение
По формуле Байеса: $\pi(p \mid n) \propto p(n∣p) \cdot \pi(p)$
$$\pi(p \mid n) \propto (\prod_{i=1}{m} p_i^{n_i})(\prod_{i=1}^{m} p_i^{\alpha_i-1}) = \prod_{i=1}^m p_i^{\alpha_i+n_i-1}$$
Таким образом, я снова получаю плотность Дирихле.  
$$p \mid n \sim Dir(a_1+n_1,...,a_m+n_m)$$
Вывод: распределение Дирихле является сопряженным к мультиномиальному правдоподобию.

# Предсказательное распределение
Усреднение Неопределенности$$P(X_{\text{new}}=i \mid \mathbf{n}) = \int p_i \, \pi(\mathbf{p} \mid \mathbf{n}) \, d\mathbf{p} = \mathbb{E}[p_i \mid \mathbf{n}]$$
$P(X_{\text{new}}=i \mid \mathbf{n})$ Предсказательная вероятность  
$\mathbb{E}[p_i \mid \mathbf{n}]$ Математическое ожидание $p_i$  
Интеграл $\int \dots d\mathbf{p}$ Усреднение по $\mathbf{p}$  
$P(X_{\text{new}}=i \mid \mathbf{n})$ — это просто Байесовская оценка самой вероятности $p_i$.
тоговая Формула (Апостериорное Среднее)  
$$P(X_{\text{new}}=i \mid \mathbf{n}) = \frac{\alpha_i + n_i}{\sum_{j=1}^m (\alpha_j + n_j)} = \frac{\alpha_i + n_i}{\alpha_0 + n}$$
Числитель $\alpha_i + n_i$ Общее количество "успехов" для исхода $A_i$, которое вы наблюдали (фиктивные + реальные).  
Знаменатель $\alpha_0 + n$ $\sum \alpha_j + \sum n_j$ Общее количество испытаний (фиктивных $\alpha_0$ + реальных $n$).
Пример (Проблема нулевых счетчиков):Вы подбросили 6-гранную кость 10 раз ($n=10$) и ни разу не выпала "шестерка" ($n_6=0$).MLE: $\hat{p}_6 = 0/10 = 0$. (Невозможно, чтобы 6-ка не выпала вообще никогда!)Байес (с неинформирующим $\alpha_i=1$): $P(X_{\text{new}}=6 \mid \mathbf{n}) = \frac{1 + 0}{6 + 10} = \frac{1}{16} \approx 0.0625$. (Разумная оценка!)Этот процесс, в частности, когда $\alpha_i=1$ (неинформирующий априор), известен как Правило преемственности Лапласа (Laplace's Rule of Succession), и он показывает силу Байесовского подхода в предотвращении экстремальных оценок.

# Серия из k новых испытаний
азвание распределения: $\mathbf{Y} \mid \mathbf{n} \sim \operatorname{Dirichlet-Multinomial}(k; \, \boldsymbol{\alpha}')$2. Функция вероятности (PMF): $P(\mathbf{Y}=\mathbf{y} \mid \mathbf{n}) = \frac{k!}{\prod y_i!} \frac{B(\boldsymbol{\alpha}' + \mathbf{y})}{B(\boldsymbol{\alpha}')}$Простыми словами, этот расчет означает: "Теперь, когда я обновил свои убеждения, какова вероятность, что в следующей серии из $k$ испытаний я получу именно $\mathbf{y} = (y_1, \dots, y_m)$ исходов?"
$$P(\mathbf{Y}=\mathbf{y} \mid \mathbf{n}) = \underbrace{\frac{k!}{\prod y_i!}}_{\text{I. Мультиномиальный Коэффициент}} \cdot \underbrace{\frac{B(\boldsymbol{\alpha}' + \mathbf{y})}{B(\boldsymbol{\alpha}')}}_{\text{II. Коэффициент Усреднения }}$$
Мультиномиальный Коэффициент:$$\frac{k!}{y_1! y_2! \cdots y_m!}$$Смысл: Это знакомый нам коэффициент, который просто подсчитывает, сколькими способами можно получить счетчики $\mathbf{y}$ в $k$ испытаниях.
$B(\boldsymbol{\alpha})$ — это Многомерная Бета-функция.
$\boldsymbol{\alpha}' = \boldsymbol{\alpha} + \mathbf{n}$ (Ваши обновленные Байесовские параметры).Числитель $B(\boldsymbol{\alpha}' + \mathbf{y})$: Многомерная Бета-функция, рассчитанная на объединенных счетчиках (старый опыт + новые данные + предсказываемые данные $\mathbf{y}$).Знаменатель $B(\boldsymbol{\alpha}')$: Многомерная Бета-функция, рассчитанная на вашем текущем знании ($\boldsymbol{\alpha}'$).Смысл: Эта дробь является результатом сложного интегрирования. Она математически выражает, насколько вероятным был бы ваш прогноз $\mathbf{y}$ относительно вашего текущего знания $\boldsymbol{\alpha}'$.  
Отличие от $P(X_{\text{new}}=i \mid \mathbf{n})$Предыдущий шаг $P(X_{\text{new}}=i \mid \mathbf{n})$ дал вам вероятность отдельного исхода (одно число).Этот шаг ($P(\mathbf{Y}=\mathbf{y} \mid \mathbf{n})$) дает вам вероятность целого вектора счетчиков (распределение).

Малое $n$ Влияние $\boldsymbol{\alpha}$ доминирует Регуляризация: Априор предотвращает экстремальные оценки (например, нулевые вероятности $p_i=0$, когда $n_i=0$), делая оценку более разумной и устойчивой.
